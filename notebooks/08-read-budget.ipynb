# 08 — Reading budget

**Question:** Does the preceding action predict how many lines the agent reads (`head -n N`, `tail -n N`)?

When an agent truncates output it writes the bound before the output is visible. On the corpus behind the companion article the median is 3 lines after a git delivery and 80 after a Python script: a factor of x27 on the same gesture.

**Kruskal-Wallis eps2** measures how much of the variance in N the preceding action type explains. A **permutation baseline** (1000 reshuffles) supplies the null: the same eps2 carries different meaning on different group structures, so it cannot be read alone.

> What is measured is the bound. "3 lines after a git commit" does not establish that the agent wanted to verify the commit — avoid that reading.

```python
SESSIONS_GLOB = "~/.openclaw/agents/*/sessions/*.jsonl"
```

```python
from agentgraph.extract import iter_shell_commands
from agentgraph.read_budget import extract_transitions, kw_epsilon2, permutation_test

records = iter_shell_commands(SESSIONS_GLOB)
transitions = extract_transitions(records)
print(f"{len(transitions)} (preceding_action, n_lines) pairs extracted")
```

```python
result = kw_epsilon2(transitions, param="n_lines", min_obs=5)
for k, v in result.items():
    print(f"{k}: {v}")
```

```python
perm = permutation_test(transitions, param="n_lines", n_perm=1000, seed=42)
for k, v in perm.items():
    print(f"{k}: {v}")
```

## Reading the result

`eps2 = 0.600` on a 295-observation, 16-transition-type corpus means the preceding action explains ~60% of the variance in N. `p_perm = 0.000` means 0 of 1000 permutations reached that score.

Your corpus will produce different numbers. The two quantities to check: is eps2 substantially above zero, and does p_perm stay below 0.05?
