# 05 — Loop inventory

**Question:** How many commands contain explicit loops — and how many are invisible to the action graph?

The transition graph built from shell segments does not emit any atom for `for` or `while` keywords — they are in `LANG_KEYWORDS` and stripped during normalisation. `xargs` and `find -exec` *are* visible because they appear as binary names.

This notebook counts both groups and checks whether loops are structurally local (prefix/loop/suffix, no nested loops in prefix or suffix).

```python
SESSIONS_GLOB = "~/.openclaw/agents/*/sessions/*.jsonl"  # adapt to your traces
```

```python
from agentgraph.extract import iter_shell_commands
from agentgraph.loops import classify_loops, invisibility, envelope_analysis

records = list(iter_shell_commands(SESSIONS_GLOB))
print(f"{len(records)} shell commands loaded")
```

```python
counts, loop_cmds = classify_loops(records)
print("Loop form counts:", counts)
print(f"\n{len(loop_cmds)} commands contain at least one explicit loop")
```

```python
inv = invisibility(counts)
for k, v in inv.items():
    print(f"{k}: {v}")
```

```python
env = envelope_analysis(loop_cmds)
for k, v in env.items():
    print(f"{k}: {v}")
```
