# 06 — SCC condensation

**Question:** How local is the feedback in your action graph?

The condensation theorem yields a DAG from any directed graph, so its existence says nothing about your corpus. Read the **size distribution of the SCCs** instead.

- All SCCs singleton *and* self-loop free: no feedback at this abstraction level. A singleton carrying a self-loop is still a cycle — read `n_cycle_sccs`, not the size distribution alone.
- One giant SCC: feedback spans the whole graph, condensation is trivial.
- Many small SCCs: feedback is local, condensation gives a useful ordering.

```python
SESSIONS_GLOB = "~/.openclaw/agents/*/sessions/*.jsonl"
```

```python
from agentgraph.extract import build_hyperedges
from agentgraph.condensation import session_transition_graph, scc_summary

edges = build_hyperedges(SESSIONS_GLOB)
print(f"{len(edges)} unique hyperedges")
```

```python
G = session_transition_graph(edges)
print(f"Transition graph: {len(G)} nodes, {sum(len(v) for v in G.values())} directed edges")
```

```python
summary = scc_summary(G)
print(summary)
print("\nSize distribution:", summary.size_distribution)
```

```python
# Is the condensation worth using?
print("Condensation is useful:", summary.condensation_is_useful())
```
