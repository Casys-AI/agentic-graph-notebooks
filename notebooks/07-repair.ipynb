# 07 — RE-PAIR grammar

**Question:** What recurring subsequences appear in your agent's action stream, and how deep do they compose?

RE-PAIR repeatedly replaces the most frequent adjacent pair with a non-terminal. The compressed grammar summarises repetition structure.

**Key output:** `depth_distribution`. Depth 0 = pair of terminal atoms. Depth N = rule whose expansion includes at least one depth-(N-1) child. A distribution concentrated at depth 0 means the agent repeats pairs, not multi-step procedures.

> Depth 0 does not mean no children: a rule can reuse another rule as a direct expansion child without increasing the depth counter.

```python
SESSIONS_GLOB = "~/.openclaw/agents/*/sessions/*.jsonl"
```

```python
from agentgraph.extract import build_hyperedges
from agentgraph.repair import build_sequence, repair, measure_grammar

edges = build_hyperedges(SESSIONS_GLOB)
seq = build_sequence(edges, normalize_sleep=True)
print(f"Sequence length: {len(seq)} symbols")
```

```python
compressed, rules = repair(seq, min_uses=3, verbose=True)
print(f"\n{len(rules)} rules, compressed length {len(compressed)}")
```

```python
report = measure_grammar(seq, compressed, rules)
for k, v in report.items():
    if k != "note":
        print(f"{k}: {v}")
print("\nNote:", report["note"])
```

## Depth-0 interpretation

A high fraction at depth 0 means the agent's sequences are dominated by *recurring pairs*, not *recurring procedures*. Depth >= 2 indicates genuine multi-level factorisation: one recurring block contains another as a proper sub-pattern.
