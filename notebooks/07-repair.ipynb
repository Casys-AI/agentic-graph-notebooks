# 07 — RE-PAIR grammar

**Question:** What recurring subsequences appear in your agent's action stream, and how deep do they compose?

RE-PAIR repeatedly replaces the most frequent adjacent pair with a non-terminal. The compressed grammar summarises the repetition structure.

`depth_distribution` carries the composition result. Depth 0 = expansion into terminal atoms only. Depth N = expansion using at least one depth-(N-1) rule, so a rule that reuses any other rule sits at depth 1 or above. A distribution concentrated at depth 0 means the recurring units are pairs of atoms; depth >= 2 means recurring blocks contain other recurring blocks.

```python
SESSIONS_GLOB = "~/.openclaw/agents/*/sessions/*.jsonl"
```

```python
from agentgraph.extract import build_hyperedges
from agentgraph.repair import build_sequence, repair, measure_grammar

edges = build_hyperedges(SESSIONS_GLOB)
seq = build_sequence(edges, normalize_sleep=True)
print(f"Sequence length: {len(seq)} symbols")
```

```python
compressed, rules = repair(seq, min_uses=3, verbose=True)
print(f"\n{len(rules)} rules, compressed length {len(compressed)}")
```

```python
report = measure_grammar(seq, compressed, rules)
for k, v in report.items():
    if k != "note":
        print(f"{k}: {v}")
print("\nNote:", report["note"])
```

## Depth-0 interpretation

A high fraction at depth 0 means the recurring units are pairs of atoms. Depth >= 2 indicates multi-level factorisation: one recurring block contains another as a proper sub-pattern.
